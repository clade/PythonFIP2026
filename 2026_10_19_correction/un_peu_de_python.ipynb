### Le try except

Lorsqu'il y a une erreur en Python, par défaut, l'interpréteur va arréter d'exécuter le code. Il est possible, en exécutant une fonction, d'empécher ce mécanisme et de gérer l'erreur. Cela se fait avec une structure ```try ... except```.

Voici un exemple : 

````python
x = None

while x is None:
    input_text = input('Entrez un nombre')
    try : 
        x = float(input_text)
    except : 
        print("Je ne peux pas convertir l'entrée en nombre")
        
````

### Créer sa propre erreur

Lorsque l'on crée une fonction, il peut être nécéssaire de créer une erreur lorsqu'il n'est pas possible de faire ce qui est demandé. Prenons un exemple : on lance une balle à la vitesse $v_0$ vers le haut, calculer le temps qu'il faut pour qu'elle atteigne le plafond situé à une hauteur $h$ du point de départ. 

Voici ce qu'il faut faire : 

````python
from math import sqrt
def temps_arrivee(h, v_0, g=9.81):
    Delta = v_0**2 - 2*g*h
    if Delta<0:
        raise Exception("La balle ne touche pas le plafond.")
    return (v_0 - sqrt(Delta))/g

print(temps_arrivee(h=1, v_0=10))
print(temps_arrivee(h=1, v_0=1))
````

- Au lieu de lever une exception, on décide d'utiliser un print de la façon suivante. Ceci est une très mauvaise idée et ne va pas résoudre le problème. Pourquoi seule la dernière ligne produit une erreur ?

````python
def temps_arrivee(h, v_0, g=9.81):
    Delta = v_0**2 - 2*g*h
    if Delta<0:
        print("La balle ne touche pas le plafond")
    else:
        return (v_0 - sqrt(Delta))/g
    
t_0 = temps_arrivee(h=1, v_0=1)
print(f"Le temps d'arrivée en seconde est {t_0} s")
print(f"Le temps d'arrivée en miliseconde est {t_0*1000} ms")
````

### Exercice 

- Créer une fonction qui calcule l'aire d'un triangle en utilisant la formule d'Héron (voir sur [wikipedia](https://fr.wikipedia.org/wiki/Formule_de_H%C3%A9ron) )

- Si l'argument de la racine carré est négatif, cela signifie que le triangle n'existe pas. Indiquer ceci par un message d'erreur explicite.

````python
from math import sqrt
def aire_triangle(a, b, c):
    p = (a + b + c)/2
    try:
        return sqrt(p*(p-a)*(p-b)*(p-c))
    except ValueError:
        raise Exception(f"Le triangle de côté {a}, {b}, {c} n'existe pas")
        
print(aire_triangle(1, 1, 1))
print(aire_triangle(10, 1, 1))
````

## Faire un graphique : diagramme de Bode

On condidère la fonction de transfert suivante : 

$$H(j\omega) = \frac{A_0}{1 + jQ\left(\frac{\omega}{\omega_0} - \frac{\omega_0}{\omega}\right)}$$

Paramètres : $\omega_0/(2\pi) = 1 $kHz, $Q=6$, $A_0=1$

On utilisera les fonctions suivantes : 

* ```subplots``` pour faire deux graphs l'un sous l'autre (avec l'options ```sharex=True```)
* ```np.angle``` pour extraire la phase d'un nombre complexe et ```np.unwrap``` pour qu'elle soit représentée de façon continue.
* ```ticker.MultipleLocator``` pour que la graduation se fasse à 45°. 
* ```grid``` avec l'argument ```which="both"``` pour pouvoir voir les graduations intermédiaire de l'échelle log.

````python
from matplotlib.pyplot import figure
import matplotlib.ticker as ticker
import numpy as np

omega_0 = 2*np.pi*1E3
omega = 2*np.pi*np.logspace(0, 6, 1001)

A_0 = 1
Q = 6

H = A_0/(1+ 1J*Q*(omega/omega_0 - omega_0/omega))

gain = np.abs(H)
phase = np.unwrap(np.angle(H))

fig = figure()
ax1, ax2 = fig.subplots(2, 1, sharex=True)

ax1.semilogx(omega/(2*np.pi), 10*np.log10(gain))
ax1.grid(which="both")
ax2.semilogx(omega/(2*np.pi), phase/(2*np.pi)*360)
ax2.yaxis.set_major_locator(ticker.MultipleLocator(45))
ax2.grid(which="both")
ax2.set_xlabel("Frequence [Hz]")
ax1.set_ylabel('Gain [dB]')
ax2.set_ylabel('Phase [deg]')
ax2.set_xlim(6, 160000)
ax1.set_ylim(-31, 1)
fig.tight_layout()
fig.savefig('bode.png')
````
