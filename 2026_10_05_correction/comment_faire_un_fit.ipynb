# Ajustement de données

Nous allons utiliser la fonction ```curve_fit``` du module ```scipy.optimize``` pour effectuer des ajustements. La façon de procéder pour un ajustement est toujours la même. 1/ on récupère les données et on les trace; 2/ on implémente le modèle dans une fonction et on cherche des paramètres de départ cohérent en traçant le modèle sur les données; 3/ on lance la fonction ```curve_fit``` pour récupérer la paramètre optimaux qui seront utilisés pour tracer la courbe en fonction à la place des paramètres initiaux et vérifier la convergence du fit.

Notons qu'il est indispendable de donner des paramètres initiaux proches des bons paramètres (sauf dans le cas d'un fit linéaire où il y a toujours convergence).

Nous allons prendre un exemple. 

* À l'aide de la fonction ```np.loadtxt``` charger le fichier ```data/data1.txt```. On utilisera l'argument ```unpack=True``` afin de créer directement deux variables ```x``` et ```y```

* Tracer les données.

````python
import numpy as np
import matplotlib.pyplot as plt
x, y = np.loadtxt('data/data1.txt', unpack=True)


plt.plot(x, y, 'o')
````

Les données peuvent être ajustées par la fonction $y(x) = y_0 + \frac{A}{1+ \left((x-x_0)/w\right)^2}$. 

* Définir une fonction python ```f(x, y_0, A, x_0, w)``` afin de modéliser les données.
* Créer un tuple ```p_0``` contenant, dans l'ordre, des valeurs initiales plausible de ```[y_0, A, x_0, w]```
* Tracer ```f``` en même temps que les données.
* Mettre une légende en utilisant l'option ```label``` de ```plot``` et la fonction ```plt.legend```.

````python
def fit_function(x,  y_0, A, x_0, w):
    return y_0 + A/(1+((x-x_0)/w)**2)

p_0 = [0, 1, -3, 1]

plt.plot(x, y, 'd', label='data')
plt.plot(x, fit_function(x, *p_0), '--', label='Initial guess')
plt.legend()
````

* En utilisant la fonction ```curve_fit``` du module ```scipy.optimize``` effectuer le fit.
* Tracer le résultat du fit sur une troisième courbe.
* Que se passe-t-il si on ne met pas l'argument ```p0``` ?
* Faire un graphique que l'on sauvegardera en pdf et qui ressemblera à ceci : mettre au moins 200 points sur les courbes, mettre des labels (```plt.xlabel``` et ```plt.ylabel```), mettre une grille de fond (```plt.grid```), sauvegarder avec ```plt.savefig```:

````python
from scipy.optimize import curve_fit

p_0 = [0, 1, -3, 1]
p_opt, p_cov = curve_fit(fit_function, x, y, p_0)

x_plot = np.linspace(-5.1, -0.9, 201)

plt.plot(x, y, 'o', label='data')
plt.plot(x_plot, fit_function(x_plot, *p_0), label='Initial guess')
plt.plot(x_plot, fit_function(x_plot, *p_opt), label='Fit')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
````
