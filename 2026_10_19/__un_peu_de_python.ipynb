## Les erreurs en Python

### Comprendre les erreurs

* Il est important de lire et comprendre les messages d'erreurs 
* C'est la dernière ligne du message qui indique la nature de l'erreur
* Python indique la ligne et l'endroit d'où vient l'erreur (sauf pour une erreur de syntaxe)

Sans executer le code, êtes vous capable de donner le nom et le message qui va apparaitre ?

### le try except

Lorsqu'il y a une erreur en python par défaut, l'interpréteur va arréter d'executer le code. Il est possible, en executant une fonction, d'empécher ce mécanisme et de gérer l'erreur. Cela se fait avec une structure ```try ... except```.

Voici un exemple : 

````python
x = None

while x is None:
    input_text = input('Entrez un nombre')
    try : 
        x = float(input_text)
    except : 
        print("Je ne peux pas convertir l'entrée en nombre")
        
````

### Créer sa propre erreur

Lorsque l'on crée une fonction, il peut être nécéssaire de créer une erreur lorsqu'il n'est pas possible de faire ce qui est demandé. Prenons un exemple : on lance une balle à la vitesse $v_0$ vers le haut, calculer le temps qu'il faut pour qu'elle atteigne le plafond situé à une hauteur $h$ du point de départ. 

Voici ce qu'il faut pas faire : 

````python
from math import sqrt
def temps_arrivee(h, v_0, g=9.81):
    Delta = v_0**2 - 2*g*h
    if Delta<0:
        raise Exception("La balle ne touche pas le plafond")
    return (v_0 - sqrt(Delta))/g

print(temps_arrivee(h=1, v_0=10))
print(temps_arrivee(h=1, v_0=1))
````
