* On considère les fonctions suivantes. Pourquoi la première fonctionne sur des tableaux et pas la seconde ? Que fait la fonction np.vectorize ? Comment l'utiliser pour résoudre le problème ? Est-ce rapide ?

````python
def gaussienne(x):
    return np.exp(-x**2/2)

def triangle(x):
    if x<-1:
        return 0.
    elif x<0:
        return 1+x
    elif x<1:
        return 1-x
    else:
        return 0.

x = np.linspace(-2, 2)
print(gaussienne(x))
print(triangle(x))
````
