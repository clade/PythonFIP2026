### Booléens

* Que vaut ```True or False and False``` ? ```(True or False) and False``` ou ```True or (False and False)``` ?
* Pourquoi seule la seconde ligne du code suivant produit une erreur ?

````python
from math import sqrt
x = -1

if x>=0 and sqrt(x)>2:
    print('x est plus grand que sqrt(2)')

if (x>=0) & (sqrt(x)>2):
    print('x est plus grand que sqrt(2)')
````

````python
# Le and est en fait un if : si le premier est faux alors c'est faux sinon, on regarde le second.
````


* Écrire une fonction ```somme``` qui prend un nombre arbitraire d'arguments et en renvoie la somme (faire la boucle explicitement)

````python
def somme(*args):
    if not args:
        return None
    s = args[0]
    for item in args[1:]:
        s += item
    return s
````

* Écrire une fonction ```primitive``` qui prend comme argument une fonction et renvoie une fonction qui est une primitive de la fonction calculée numériquement. Le nombre de pas sera une argument optionnel. On prendra un algorithme quelconque (ce n'est pas le problème). On doit pouvoir faire :

````python
def f(x):
    return 1/(1+x**2)
F = primitive(f)
print(F(1))

F = primitive(f, N=10000)
print(F(1))
````

````python
import numpy as np

def primitive(f, N=1000):
    def primitive_f(x):
        Tx = np.linspace(0, x, N, endpoint=False)
        return np.trapezoid(f(Tx), Tx)
    return np.vectorize(primitive_f)

primitive(np.sin)(1)
````
